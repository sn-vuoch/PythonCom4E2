## Exercise 3: Operator precedence check

- Create a complex boolean expression using `and`, `or`, and `not`.
- Evaluate it once without extra parentheses and once with explicit parentheses.
- Print both and comment whether they match.

```python
# Write your solution here
p = True
q = False
r = True

# rely on Python's default operator precedence
implicit = p or q and not r

# spell out the grouping explicitly
explicit = p or (q and (not r))

print("Implicit (no extra parentheses):", implicit)
print("Explicit (fully parenthesize): ", explicit)
print("Do they agree?                 ", implicit == explicit)
```

## Exercise 4: Identity vs equality

- Create two equal lists and compare with `==` and `is`.
- Assign one list to another variable and compare again.
- Write a short comment explaining differences in output.

```python
# Write your solution here
fruits = ['orange', 'banana', 'mango']
new_fruits = ['orange', 'banana', 'mango']
print(fruits == new_fruits)
print(fruits is new_fruits)

new_fruits = fruits
print(fruits == new_fruits)
print(fruits is new_fruits)

# For ==, Python compares values in the list.
# For is, Python compares memory location.
```

## Exercise 5: Membership filter
